# Phase 7 F1 — public aggregate results

This notebook presents privacy-safe aggregate F1 results. It contains no audio, manifests, IDs, prediction-level outputs, model weights, or test-set results.

```python
from pathlib import Path
from IPython.display import display
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path('..')
metrics = pd.read_csv(ROOT / 'docs' / 'PHASE7_F1_AGGREGATE.csv')
display(metrics)
print('Best F1 epoch:', int(metrics.loc[metrics['CER_%'].idxmin(), 'epoch']))
print('B0 CER: 103.12924%')
print('Test data used: False')
```

## Interpretation

F1 reduced validation CER from the B0 value of 103.13% to 51.35% at Epoch 3. Training loss continued to decrease after Epoch 3 while validation CER worsened, so Epoch 3 is selected by validation CER and later epochs are not treated as improvements. A validation-only repetition-penalty ablation reached 48.55%; it remains a development result until the final sealed test evaluation.
